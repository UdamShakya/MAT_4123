# Genetic Algorithm for Two Varibales 

### Initialization


```python
import random 

BIT_SIZE =6
total_bits = BIT_SIZE * 2
POP_SIZE =6
PC= 0.85
PM =0.2
MAX_GEN =5

XL =0
XU= 2

STEP_SIZE = (XU - XL) / (2 ** BIT_SIZE - 1)

def Himmelblau(x):
    x1,x2=x

    return (x1**2 + x2 -11)**2 + (x1 + x2**2 -7)**2


def decode(chromosone):
    x1_bits= chromosone[:BIT_SIZE]
    x2_bits= chromosone[BIT_SIZE:]
    s1=int(x1_bits,2)
    s2=int(x2_bits,2)
    x1= XL + s1 * STEP_SIZE
    x2= XL + s2 * STEP_SIZE
    return (x1, x2)

population=[]

for i in range(POP_SIZE):
    Chromosone=""
    for j in range(total_bits):
        Chromosone+=random.choice("01")
    population.append(Chromosone)
```

## Genetic Algorithm loop 

```python
generation = 0 

while generation< MAX_GEN:
     print("\n generation", generation)

     fitness_Values=[]
     total_fitness=0

     #evaluation
     for chrom in population:
          x1, x2 = decode(chrom)
          x=x1,x2
          fitness = Himmelblau((x))
          fitness_Values.append(1/(1+fitness))
          total_fitness += 1/(1+fitness)

          print("chromosone:", chrom, "x:", x, "fitness:", fitness)

     #selection
     probabilities=[]
     cumulative_prob=[]
     cumulative = 0

     for fitness in fitness_Values:
          prob=fitness/total_fitness
          probabilities.append(prob)
          cumulative += prob
          cumulative_prob.append(cumulative)

     mating_pool=[]

     for i in range (POP_SIZE):
          r=random.random()
          for j in range(POP_SIZE):
               if r <= cumulative_prob[j]:
                    mating_pool.append(population[j])
                    break

     #crossover
     random.shuffle(mating_pool)
     new_population=[]

     for i in range (0,POP_SIZE,2):
          parent1=mating_pool[i]
          parent2=mating_pool[i+1]

          r= random.random()

          if r<PC:
               point=3 
               child1=parent1[:point] + parent2[point:]
               child2=parent2[:point] + parent1[point:]
          else:
               child1=parent1
               child2=parent2

          new_population.append(child1)
          new_population.append(child2)

     #mutation

     final_population=[]

     for chrom in new_population:
          new_chrome="" 
          for bit in chrom:
               r=random.random()
               if r<PM:
                    if bit=="0":
                         new_bit="1"
                    else:
                         new_bit="0"
               else:
                   new_bit=bit
               new_chrome+=new_bit
          final_population.append(new_chrome)

     population=final_population
     generation += 1
```

### Final Result

```python
best_value=99999
best_solution=""

for chrom in population:
    x1, x2 = decode(chrom)
    x = x1, x2
    fitness = Himmelblau((x))
    if fitness < best_value:
        best_value = fitness
        best_solution = chrom

print("\nBest solution:", best_solution, "Decoded:", decode(best_solution), "Fitness:", best_value)
```
