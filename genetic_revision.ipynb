```python
import random

pop_size=6
bit_length=6
pc=0.8
pm=0.2
max_gen=5

xl=0
xu=2

step_size=(xu-xl)/(2**bit_length-1)

def f(x):
    return -x**2+ 2*x


def decode(chromosome):
    decimal_value=int(chromosome,2)
    x = xl + decimal_value * step_size
    return x


population=[]

for i in range(pop_size):
    chromosome="" 
    for j in range(bit_length):
        chromosome+=random.choice("01")
    population.append(chromosome)

print("initial Population",population)



generation=0

while generation<max_gen:

    print("\n generation:",generation)

    fitness_values=[]
    total_fitness=0

    for chrom in population:
        x=decode(chrom)
        fitness= f(x)
        fitness_values.append(fitness)
        total_fitness+=fitness

        print(f"chromosome: {chrom}, x: {x:.4f}, fitness: {fitness:.4f}")

    print("roullete wheel method")

    probabilities=[]
    cumulative_prob=[]

    cumulative=0

    for fitness in fitness_values:
        prob=fitness/total_fitness
        probabilities.append(prob)
        cumulative+=prob
        cumulative_prob.append(cumulative)

    for i in range(pop_size):
        print(f"chromosome: {population[i]}, fitness: {fitness_values[i]:.4f}, probability: {probabilities[i]:.4f}, cumulative probability: {cumulative_prob[i]:.4f}")

    mating_pool=[]

    for i in range(pop_size):
        r=random.random()
        print(f"random number :{r:.4f}")
        for j in range(pop_size):
            if r < cumulative_prob[j]:
                mating_pool.append(population[j])
                break


    new_population=[]
    random.shuffle(mating_pool)
    for i in range(0,pop_size,2):
        parent1=mating_pool[i]
        parent2=mating_pool[i+1]

        r=random.random()

        print(f"Parents: {parent1}, {parent2}")

        if r < pc:
            point=3
            child1=parent1[:point]+parent2[point:]
            child2=parent2[:point]+parent1[point:]
        else:
            child1=parent1
            child2=parent2

        print(f"Children: {child1}, {child2}")

        new_population.extend([child1, child2])

    final_population=[]

    for chrom in new_population:
        new_chrom="" 
        i=1
        for bit in chrom:
            r=random.random()
            
            if r<pm:
                if bit=="0":
                    new_bit="1"
                else:
                    new_bit="0"
            else:
                new_bit=bit

            new_chrom+=new_bit

        final_population.append(new_chrom)

    population=final_population
    generation+=1


best_fitness=-9999
best_solution=""

for chrom in population:
    x=decode(chrom)
    fitness=f(x)

    if fitness>best_fitness:
        best_fitness=fitness
        best_solution=chrom

print(f"Best solution: {best_solution}, Fitness: {best_fitness}")
```

```python
import random

pop_size=10
bitlen=6
total_bit=12
pc=0.85
pm=0.2
max=10
xl=0
xu=10

step_size=(xu-xl)/(2**bitlen-1)

def f(x1, x2):
    return (x1**2 + x2 - 11)**2 + (x1 + x2**2 - 7)**2

def decode(chromosome):
    x1_chrom=chromosome[:bitlen]
    x2_chrom=chromosome[bitlen:]

    s1=int(x1_chrom,2)
    s2=int(x2_chrom,2)

    x1=xl + s1*step_size
    x2=xl + s2*step_size

    return x1,x2



population=[]
for i in range(pop_size):
    chromosome=""
    for j in range(total_bit):
        chromosome+=random.choice("01")
    
    population.append(chromosome)

geneartion=0

while geneartion<max_gen:

    fitness_values=[]
    total_fitness=0

    bestvalue=99999
    best_chrome=""

    for chrom in population:
        x1,x2=decode(chrom)
        fitness=1+(1+f(x1,x2))
        fitness_values.append(fitness)
        total_fitness+=fitness

        if f(x1,x2)<bestvalue:
            bestvalue=fitness
            best_chrome=chrom

        print(f"chromosome: {chrom}, x1: {x1:.4f}, x2: {x2:.4f}, fitness: {fitness:.4f}")


    probabilities=[]
    cumulative_prob=[]
    cumulative=0

    for fitness in fitness_values:
        prob=fitness/total_fitness
        probabilities.append(prob)
        cumulative+=prob
        cumulative_prob.append(cumulative)

    mating_pool=[]
    for i in range(pop_size):
        r=random.random()
        for j in range(pop_size):
            if r<cumulative_prob[j]:
                mating_pool.append(population[j])
                break

    random.shuffle(mating_pool)
    new_population=[]

    for i in range(0,pop_size,2):
        parent1=mating_pool[i]
        parent2=mating_pool[i+1]

        r=random.random()

        if r<pc:
            point=3
            child1=parent1[:point]+parent2[point:]
            child2=parent2[:point]+parent1[point:]
        else:
            child1=parent1
            child2=parent2

        new_population.extend([child1,child2])


    final_population=[]
    for chrom in new_population:
        chrom=""
        r=random.random()

        for bit in chrom:
            if r<pm:
                if bit=="1":
                    new_bit="0"
                else:
                    new_bit="1"
            else:
                new_bit="0"

            chrom+=new_bit

    final_population[0]=best_chrome
    population=final_population
    generation+=1










```

```python
import random 
import math 

def f(x):
    return x**4 -3*x**2 +x+5

x=1.5
T=8
epsilon=1e-5
n=2
alpha=0.5
t=0

lower=-2
upper=2

sigma=(upper-lower)/6

while T >1e-6:
    step=random.gauss(0, sigma)
    x_new=max(lower,min(upper,x+step))

    deltaE=f(x_new)-f(x)

    if deltaE<0:
        if abs(x_new-x)<epsilon:
            break

        x=x_new
        t+=1

    else:
        r=random.random()
        if r<=math.exp(-deltaE/T):

            if(abs(x_new-x)<epsilon):
                break

            x=x_new
            t+=1

    if t%n==0 and t!=0:
        T=alpha*T

print("minimum approx at x =",x)
print("f(x) at minimum approx =",f(x))
            
```
