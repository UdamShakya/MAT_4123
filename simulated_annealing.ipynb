# Simulated Annealing

## Single Variable

```python
import random
import math 

def f(x):
    return x**4 - 3*x**2 + x+5

x=1.5
T=8
epsilon=1e-4
n=2
alpha=0.5
t=0

lower =-2
upper =2
sigma = (upper-lower)/6

while T > 1e-6:
    step=random.gauss(0,sigma)
    x_new=max(lower, min(upper, x+step))

    deltaE=f(x_new)-f(x)

    if deltaE<0:
        if abs(x_new-x) < epsilon:
            break

        x=x_new
        t+=1
    else:
        r=random.random()
        if r<=math.exp(-deltaE/T):
            if abs(x_new-x) < epsilon:
                break

            x=x_new
            t+=1

    if t%n==0 and t!=0:
        T=alpha*T

print("minimum approx at x= ",x)
print("f(x)= ",f(x))
```

## Two variable 

```python
import random 
import math 

def f(x1,x2):
    return (x1**2+x2-11)**2+(x1+x2**2-7)**2

x1 ,x2 = 2.5,2.5
T=405
epsilon =1e-3
n=10
alpha=0.6
t=0

lower =0 
upper=5


sigma= (upper-lower)/6

while T >1e-6:

    step1= random.gauss(0,sigma)
    step2=random.gauss(0,sigma)

    x1_new=max(lower, min(upper, x1+step1))
    x2_new=max(lower,min(upper,x2+step2))

    deltaE = f(x1_new,x2_new) - f(x1,x2)

    if deltaE<0:

        if math.sqrt((x1_new-x1)**2+(x2_new-x2)**2) < epsilon:
            break

        x1,x2 = x1_new,x2_new
        t+=1

    else:
        r=random.random()
        if r<=math.exp(-deltaE/T):
            if math.sqrt((x1_new-x1)**2+(x2_new-x2)**2) < epsilon:
                break

            x1,x2 = x1_new,x2_new
            t+=1

    if t%n==0 and t!=0:
        T*=alpha

print("Approximate minimum at:")
print("x1= ",x1)
print("x2= ",x2)
print("f(x1,x2)= ",f(x1,x2))
```
