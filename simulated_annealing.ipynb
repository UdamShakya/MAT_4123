# Simulated Annealing

## Single Variable

```python
import random
import math 

def f(x):
    return x**4 - 3*x**2 + x+5

x=1.5
T=8
epsilon=1e-4
n=2
alpha=0.5
t=0

lower =-2
upper =2
sigma = (upper-lower)/6

while T > 1e-6:
    step=random.gauss(0,sigma)
    x_new=max(lower, min(upper, x+step))

    deltaE=f(x_new)-f(x)

    if deltaE<0:
        if abs(x_new-x) < epsilon:
            break

        x=x_new
        t+=1
    else:
        r=random.random()
        if r<=math.exp(-deltaE/T):
            if abs(x_new-x) < epsilon:
                break

            x=x_new
            t+=1

    if t%n==0 and t!=0:
        T=alpha*T

print("minimum approx at x= ",x)
print("f(x)= ",f(x))
```
