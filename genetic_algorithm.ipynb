# Genetic Algorithm 

## Single variable 

```python
import random

POP_SIZE=6
BIT_LENGTH= 6
PC=0.85
PM=0.20
MAX_GEN = 5 

XL=0
XU=2

STEP_SIZE= (XU-XL)/(2**BIT_LENGTH-1)

def f(x):
    return -x**2 + 2*x


def decode(chromosome):
    decimal_value= int(chromosome, 2)
    x= XL + decimal_value*STEP_SIZE
    return x

population=[]

for i in range(POP_SIZE):
    chromosome=""
    for j in range(BIT_LENGTH):
        chromosome+=random.choice("01")
    population.append(chromosome)


print("Initial Population:",population)
```

## GA MAIN LOOP 

```python
generation =0 

while generation < MAX_GEN:
    print("\n---------------")
    print("Generation:", generation)

#1.FITNESS EVALUATION

    fitness_values =[]
    total_fitness=0

    print("Evalauation")
    for chrom in population:
        x= decode(chrom)
        fitness=f(x)
        fitness_values.append(fitness)
        total_fitness+=fitness

        print("chromosome:" , chrom,
              "x=",round(x,4),
              "fitness=",round(fitness,4)
        )

# 2. SELECTION/REPRODUCTION

    print("\n Selection : Roulette Wheel Method): ")

    probabilities = []
    cumulative_prob = []
    cumulative = 0

    for fitness in fitness_values:
        prob = fitness / total_fitness
        probabilities.append(prob)
        cumulative += prob
        cumulative_prob.append(cumulative)

    for i in range(POP_SIZE):
        print("chromosome:", population[i],
            "fitness:", round(fitness_values[i],4),
            "probability:", round(probabilities[i],4),
            "cumulative probability:", round(cumulative_prob[i],4)
        )

        mating_pool =[]

        for i in range (POP_SIZE):
            r = random.random()
            for j in range(POP_SIZE):
                if r <= cumulative_prob[j]:
                    mating_pool.append(population[j])
                    break
    # 3. CROSSOVER

        print ("\n Crossover: ")

        new_population= []
        random.shuffle(mating_pool)
        for i in range(0, POP_SIZE, 2):
            parent1 = mating_pool[i]
            parent2 = mating_pool[i+1]

            print("Parent 1:", parent1, "Parent 2:", parent2)

            if random.random() < PC:
                point = random.randint(1, BIT_LENGTH-1)
                child1 = parent1[:point] + parent2[point:]
                child2 = parent2[:point] + parent1[point:]
                print("Crossover happened")

                print("Crossover Point:", point)
                print("Child 1:", child1, "Child 2:", child2)
            else:
                child1, child2 = parent1, parent2
                print("No Crossover. Children are same as parents.")

            new_population.extend([child1, child2])

    #4.MUTATION

    print("\n mutation")
    final_population=[]

    for chrom in new_population:
        new_chrome=""
        print("Original Chromosome:", chrom)
        print("---------------")
        i=1
        for bit in chrom:
            print("consider bit ",i)
            r= random.random()
            if r< PM :
                if bit == "0":
                    newbit="1"
                else:
                    newbit="0"
                print("bit flipped in position:", i," in chrom",chrom)
            else:
                newbit=bit
            new_chrome+=newbit
            i+=1
        final_population.append(new_chrome)
        print("Mutated Chromosome:", new_chrome)

    population= final_population
    generation += 1



```

## Final Result 

```python
print("final population")

best_fitness= -999
best_solution=""

for chrom in population:
    x=decode(chrom)
    fitness=f(x)

    print("chromosome:", chrom,
          "x=",round(x,4),
          "fitness=",round(fitness,4)
    )

    if fitness > best_fitness:
        best_fitness = fitness
        best_solution = chrom

print("\nBest Solution:")
print("Chromosome:", best_solution)
print("x=", round(decode(best_solution),4))
print("Maximum f(x)", best_fitness) 
```
